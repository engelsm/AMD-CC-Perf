# Stream Benchmark Analysis

This notebook loads benchmark results from `stream_out.csv`, compares performance across environments, plots best rates by function, and calculates overhead relative to the native environment.

```python
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt
import os
import sys

sys.path.append("..")
import common_style as cs

cs.apply_style()

os.makedirs("images", exist_ok=True)

df = pd.read_csv("../../outputs/stream/results.csv")

fig, ax = plt.subplots(figsize=(cs.COLW, 0.85))
sns.barplot(
    data=df, x="Function", y="BestRate_MBs", hue="Env",
    hue_order=cs.MODE_ORDER, palette=cs.MODE_PALETTE,
    edgecolor="black", linewidth=0.6, ax=ax,
)
ax.set_ylim(df["BestRate_MBs"].min() * 0.9)
ax.set_ylabel("Best Rate (MB/s)")
ax.set_xlabel("")
cs.top_legend(fig, ax, ncol=3)
fig.savefig("images/stream_plot.pdf")

df_native = df[df["Env"] == "Native"].set_index("Function")
df["Overhead_pct"] = df.apply(lambda row: (1 - row["BestRate_MBs"] / df_native.loc[row["Function"], "BestRate_MBs"]) * 100, axis=1)
print(df)
df.groupby("Env")["Overhead_pct"].mean()
```
